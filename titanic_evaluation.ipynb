# Q. find probability of survival for femaes in 1st class

```python
var=df[df["Pclass"]==1] # vfirstly filter 1st p class 
var1=var[var["Sex"]=="female"] # after that filter by gender
f=var1['PassengerId'].count()
```

```python
survived = var1["Survived"].sum()
p=survived/f
print("probability of survival for females in 1st class:",p)
print("rate of survival",p*100,"%")
```

# Q. Detect outliers in Fare

```python
Q1=df["Fare"].quantile(0.25)
print(Q1)
Q3=df["Fare"].quantile(0.75)
print(Q3)

```

```python
IQR=Q3-Q1
IQR.item()
```

```python
upper=Q3+1.5*IQR
print(upper)
```

```python
lower=Q1-1.5*IQR
print(lower)
```

```python
outliers = df[
    (df["Fare"] < lower) |
    (df["Fare"] > upper)
]
print(outliers)
```

```python
print(outliers[[ "Fare"]].round(2))
```

# Q. create a function to categorize passengers by fare range

```python
def fare_category(Fare):
     if Fare<50:
         return "low"
     elif Fare==50:
         return "medium"
     elif Fare>50:
         return "high"
     else:
         return "very high"
df["Fare_range"]=df["Fare"].apply(fare_category)
print(df)
```
